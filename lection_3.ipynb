```python
def get_type_val(value):
    match value:
        case int(value):
            print("it's an integer")
        case str(value):
            print("it's a string")
        case _:
            print("It's another type")

get_type_val(10)
get_type_val('10')
get_type_val(10.2)
```

```python
def calc(a, b, operation):
    match operation:
        case "+" | "plus":
            return a + b
        case "-" | "minus":
            return a - b
        case "*" | "multiply":
            return a * b
        case "/" | "divide":
            return a / b if b != 0 else "Делить на ноль нельзя!"

print(calc(10, 7, 'divide'))
print(calc(10, 0, 'divide'))
print(calc(10, 7, 'plus'))
print(calc(10, 7, 'minus'))
print(calc(10, 7, 'multiply'))
```
